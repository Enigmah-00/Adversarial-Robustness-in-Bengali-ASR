# NB-23 — the same weights, attacked a second time

**What this is for.** The supervisor review's first point: sigma = 21.6 pp is
called run-to-run *training* variance, but every checkpoint was attacked once,
so it also contains the variance of the attack itself. We have exactly one pair
of attacks on identical weights (41.0% and 61.0%), which is one degree of
freedom and settles nothing.

This notebook re-attacks checkpoints that are **already measured**. Each one
becomes a second measurement of the same weights, so the spread between the two
is attack variance with the training held exactly fixed.

**What actually makes the two runs differ, and it is not the seed.** The attack
draws from no random number generator: `delta` is initialised to zeros, there
are no random restarts, the model is in eval mode with SpecAugment off, and the
batches are duration-sorted rather than shuffled. `ATTACK_SEED` is therefore a
run LABEL, not a mechanism --- setting it changes nothing in the optimisation.

The two runs differ because of GPU nondeterminism: `cudnn.benchmark = True`
re-selects convolution kernels per session, and `USE_AMP = True` makes fp16
accumulation order vary. That is deliberate and it is the right comparison,
because it is exactly the mechanism behind the pair we already have --- RUNLOG
records the cause of the 41.0% / 61.0% gap on 100h-final T5 as
`cudnn.benchmark`. Introducing a random restart instead would make this run
differ from the published one in CONFIGURATION as well as in draw, and the gap
would stop being comparable.

So report what this measures as **attack-measurement variance**, never as
"attack-seed variance". The distinction matters: a seed-driven spread would be
removable by fixing the seed, and this one is not --- it is removable only by
`torch.use_deterministic_algorithms(True)`, which the published runs did not
use. See NB-18 for the determinism control.

**One honest risk.** Because nothing is re-drawn deliberately, the re-run may
land close to the published figure. A small gap is a real result too: it would
make the 20 pp pair look like an outlier and push sigma back toward training
variance. Either outcome resolves the open question.

**Three pairs instead of one.** The queue re-attacks one checkpoint at each of
the 1, 5 and 25-hour rungs, all three on `transformers` 5.16.1, all three
already measured in NB-16. That gives attack variance on 3 degrees of freedom,
against the 1 it rests on now.

**What changes versus NB-21, and what does not.**

| | |
|---|---|
| attack configuration | **identical** — same eps range, bisection rounds, step counts, batch size, LR |
| the weights | **identical** — the same checkpoint directories NB-16 attacked |
| eval set and targets | **identical** — the same 100 utterances, same SHA-256-checked target file |
| **attack seed** | set, but it feeds nothing --- see above |
| **what moves** | **GPU nondeterminism: cudnn kernel choice and fp16 accumulation order** |
| per-probe logging | **new** — steps used and why each probe stopped |

**Output names are deliberately different.** These write
`attackrep_{tag}_a{seed}_T5_n100.csv`, not `eps50_*`. The existing analysis
globs for `eps50_*`; if these files matched that pattern they would be pooled in
as if they were new training runs, which is the opposite of what they measure.

**Cost.** About 4.6–6.3 GPU-hours per condition, three conditions. One or two
per session at `WALL_HOURS = 11`. The resume path carries a part-finished
condition across sessions — publish the output, attach it back, re-commit.

## 4 · Front end, loss, PGD, bisection

Byte-identical to NB-05. These read `model`, `processor`, `BLANK_ID` and `DO_NORMALIZE` as globals at call time, which is what lets Stage 1 swap checkpoints without redefining anything.

```python
SR = 16000

def load_batch(rows):
    """Length-bucketed batch. Returns padded waveforms + a validity mask."""
    wavs = []
    for _, r in rows.iterrows():
        a, _ = librosa.load(r["path"], sr=SR, mono=True)
        a = np.asarray(a, np.float32)
        pk = np.abs(a).max()
        wavs.append(a / max(pk, 1.0))
    T = max(len(w) for w in wavs)
    x = torch.zeros(len(wavs), T, dtype=torch.float32)
    m = torch.zeros(len(wavs), T, dtype=torch.float32)
    for i, w in enumerate(wavs):
        x[i, :len(w)] = torch.from_numpy(w); m[i, :len(w)] = 1.0
    return x.to(DEVICE), m.to(DEVICE)


def frontend_b(x, mask):
    """Masked zero-mean/unit-variance. Computing stats over the PADDING would
    corrupt short utterances in a mixed-length batch."""
    if DO_NORMALIZE:
        n = mask.sum(-1, keepdim=True).clamp(min=1)
        mu = (x * mask).sum(-1, keepdim=True) / n
        var = (((x - mu) * mask) ** 2).sum(-1, keepdim=True) / n
        x = (x - mu) / torch.sqrt(var + 1e-7)
        x = x * mask
    return x


def logits_b(x, mask):
    return model(input_values=frontend_b(x, mask),
                 attention_mask=mask.long()).logits


def ctc_per_sample(x, mask, tgt_pad, tgt_len):
    """Per-sample CTC loss -> lets us mask out finished samples."""
    lg = logits_b(x, mask)
    logp = F.log_softmax(lg.float(), -1).transpose(0, 1)          # (L,B,V)
    in_len = model._get_feat_extract_output_lengths(mask.sum(-1).long()).long()
    in_len = in_len.clamp(max=logp.shape[0])
    return F.ctc_loss(logp, tgt_pad, in_len, tgt_len,
                      blank=BLANK_ID, zero_infinity=True, reduction="none")


def decode_b(x, mask):
    with torch.no_grad():
        ids = torch.argmax(logits_b(x, mask), -1)
    return [norm_bn(s) for s in processor.batch_decode(ids)]


def int16_rt(x):
    return torch.clamp((x * 32767.0).round(), -32768.0, 32767.0) / 32767.0

def snr_db_b(clean, adv, mask):
    n = (adv - clean) * mask
    s = clean * mask
    return (10 * torch.log10(s.pow(2).sum(-1) / (n.pow(2).sum(-1) + 1e-12))).cpu().numpy()
```

```python
def pgd_batch(x, mask, tgt_pad, tgt_len, targets, eps_vec,
              steps, delta_init=None, lr=LR):
    """eps_vec: (B,) per-sample L-inf budgets.
       Returns (success bool array, frozen deltas, steps_used)."""
    B = x.shape[0]
    eps_col = eps_vec.view(-1, 1)
    delta = (torch.zeros_like(x) if delta_init is None
             else torch.max(torch.min(delta_init.clone(), eps_col), -eps_col))
    delta.requires_grad_(True)
    opt = torch.optim.Adam([delta], lr=lr)
    scaler = torch.amp.GradScaler("cuda", enabled=USE_AMP)

    active = torch.ones(B, dtype=torch.bool, device=DEVICE)
    frozen = torch.zeros_like(x)
    success = np.zeros(B, dtype=bool)
    best_loss = torch.full((B,), float("inf"), device=DEVICE)
    stall = torch.zeros(B, dtype=torch.long, device=DEVICE)

    for step in range(steps):
        if not active.any():
            break
        x_adv = torch.clamp(x + delta, -1, 1) * mask
        with torch.amp.autocast("cuda", enabled=USE_AMP):
            per = ctc_per_sample(x_adv, mask, tgt_pad, tgt_len)
        loss = (per * active.float()).sum() / active.float().sum().clamp(min=1)

        opt.zero_grad(); scaler.scale(loss).backward()
        scaler.unscale_(opt)
        if delta.grad is not None:
            delta.grad[~active] = 0          # finished samples stop moving
        scaler.step(opt); scaler.update()
        with torch.no_grad():
            delta.data = torch.max(torch.min(delta.data, eps_col), -eps_col)

        with torch.no_grad():                # stall detection
            improved = per < best_loss - 1e-3
            best_loss = torch.minimum(best_loss, per.detach())
            stall = torch.where(improved, torch.zeros_like(stall), stall + 1)
            give_up = active & (stall > STALL_PATIENCE)
            active = active & ~give_up

        if (step + 1) % CHECK_EVERY == 0:
            hyp = decode_b(int16_rt(torch.clamp(x + delta, -1, 1)) * mask, mask)
            with torch.no_grad():
                for i in range(B):
                    if active[i] and hyp[i] == targets[i]:
                        frozen[i] = delta[i].detach()
                        success[i] = True
                        active[i] = False
    return success, frozen, step + 1
```

```python
def eps50_batch(x, mask, tgt_pad, tgt_len, targets, verbose=True):
    """Per-sample geometric bisection for the smallest budget that still works.

    ---------------------------------------------------------------------------
    BUG FIXED 2026-09-22 (found in the rung-1 T5 run, 40% censoring).

    The previous version ended with

        ok, best, _ = pgd_batch(..., hi, CONFIRM_STEPS, delta_init=best)

    which REASSIGNED `best` wholesale to pgd_batch's return value. That return
    is a `frozen` buffer holding zeros for any sample that did not succeed on
    that particular pass -- so one unlucky final pass wiped out a delta that had
    already been verified at a larger budget. Those samples were then written
    out as censored, with SNR 70-86 dB (the signature of delta == 0) even though
    the ceiling probe had already proved them attackable.

    In the rung-1 run that discarded ~6 of 25 real measurements. Worse, it does
    so at a RATE THAT VARIES BY MODEL, so comparing median eps50 across rungs
    would have been comparing differently-biased subsamples -- straight into
    RQ1's headline.

    Fix: keep a running `best_eps` / `best_delta` that is only ever updated on a
    VERIFIED success. Nothing overwrites a confirmed result. Censoring now means
    exactly one thing: the attack failed at every budget tried, ceiling included.

    Side benefit: eps50 becomes a guaranteed UPPER BOUND on the true minimum.
    Optimiser noise can only make it conservative, never wrong in an
    unpredictable direction -- which is the property you want in a metric you
    are about to compare across four models.
    ---------------------------------------------------------------------------
    """
    B = x.shape[0]
    lo = torch.full((B,), EPS_LO, device=DEVICE)
    hi = torch.full((B,), EPS_HI, device=DEVICE)
    best_eps   = torch.full((B,), float("nan"), device=DEVICE)   # verified only
    best_delta = torch.zeros_like(x)

    def record(okt, eps_vec, d):
        """Adopt a result only when it is both successful and better."""
        nonlocal best_eps, best_delta
        better = okt & (torch.isnan(best_eps) | (eps_vec < best_eps))
        best_eps   = torch.where(better, eps_vec, best_eps)
        best_delta = torch.where(better.view(-1, 1), d, best_delta)

    # Ceiling probe. Failure here is genuine right-censoring.
    ok, d, _ = pgd_batch(x, mask, tgt_pad, tgt_len, targets, hi, CONFIRM_STEPS)
    okt = torch.tensor(ok, device=DEVICE)
    record(okt, hi, d)
    feasible = ok.copy()
    if verbose:
        print(f"  ceiling probe eps={EPS_HI:.1e}: {feasible.sum()}/{B} feasible, "
              f"{(~feasible).sum()} censored")

    for rnd in range(BISECT_ROUNDS):
        mid = torch.sqrt(lo * hi)
        ok, d, _ = pgd_batch(x, mask, tgt_pad, tgt_len, targets, mid,
                             PROBE_STEPS, delta_init=best_delta)
        ok = ok & feasible
        okt = torch.tensor(ok, device=DEVICE)
        hi = torch.where(okt, mid, hi)
        lo = torch.where(okt, lo, mid)
        record(okt, mid, d)
        if verbose:
            print(f"  round {rnd+1}/{BISECT_ROUNDS}: {ok.sum()}/{B} succeeded "
                  f"| median eps {mid.median().item():.3e}")

    # Budget-consistency round: one more bisection at the FULL step budget, so
    # the reported boundary is measured at the budget eps50 is defined at rather
    # than at the cheaper probe budget. See RUNLOG for why this matters.
    #
    # Skipped in FIXED_EPS mode: there is no bracket to refine. The ceiling probe
    # already ran at CONFIRM_STEPS, so the single reported budget is measured at
    # the full budget and needs no second pass.
    if BISECT_ROUNDS == 0:
        eps       = best_eps.cpu().numpy()
        confirmed = ~np.isnan(eps)
        if verbose:
            print(f"  -> FIXED_EPS {EPS_HI:.1e}: {confirmed.sum()}/{B} succeeded, "
                  f"{(~confirmed).sum()} resisted")
        return eps, np.ones(B), confirmed, best_delta

    mid = torch.sqrt(lo * hi)
    ok, d, _ = pgd_batch(x, mask, tgt_pad, tgt_len, targets, mid,
                         CONFIRM_STEPS, delta_init=best_delta)
    ok = ok & feasible
    okt = torch.tensor(ok, device=DEVICE)
    record(okt, mid, d)
    if verbose:
        print(f"  budget-consistency round @ {CONFIRM_STEPS} steps: {ok.sum()}/{B}")

    eps       = best_eps.cpu().numpy()
    confirmed = ~np.isnan(eps)                 # censored == never succeeded
    width     = (hi / lo).cpu().numpy()
    if verbose:
        print(f"  -> {confirmed.sum()}/{B} measured, {(~confirmed).sum()} censored")
    return eps, width, confirmed, best_delta
```

## 5 · Per-condition loader

```python
def load_condition(cfg):
    """Resolve and load one checkpoint, rebinding the globals the attack closures read.

    The closures in the cells above look up `model`, `processor`, `BLANK_ID` and
    `DO_NORMALIZE` at CALL time, not definition time, so rebinding them here is
    all that is needed to point the whole machine at a different checkpoint.
    Every one of the four must be rebound: leaving a stale BLANK_ID or
    DO_NORMALIZE behind would silently attack the new model through the old
    model's decoding assumptions.
    """
    global model, processor, BLANK_ID, DO_NORMALIZE, MODEL_DIR, RUNG, CHECKPOINT
    ds = cfg["ckpt_dataset"]
    cands = [p for p in inputs if re.search(rf"{re.escape(ds)}$", os.path.basename(p))]
    assert len(cands) == 1, (
        f"Need exactly one input matching '{ds}', found {len(cands)}: "
        f"{[os.path.basename(c) for c in cands]}")
    ck = cands[0]
    if cfg["checkpoint"] == "final":
        MODEL_DIR = ck if os.path.exists(f"{ck}/config.json") else f"{ck}/final"
    else:
        MODEL_DIR = f"{ck}/milestones/step-{cfg['checkpoint'].split('-')[1]}"
    assert os.path.exists(f"{MODEL_DIR}/config.json"), f"no config.json under {MODEL_DIR}"

    processor = Wav2Vec2Processor.from_pretrained(MODEL_DIR)
    model = Wav2Vec2ForCTC.from_pretrained(MODEL_DIR).float().to(DEVICE).eval()
    for p in model.parameters():
        p.requires_grad_(False)
    model.config.apply_spec_augment = False
    DO_NORMALIZE = bool(getattr(processor.feature_extractor, "do_normalize", True))
    BLANK_ID = model.config.pad_token_id
    RUNG, CHECKPOINT = cfg["rung"], cfg["checkpoint"]
    print(f"  loaded {MODEL_DIR}")
    print(f"  do_normalize {DO_NORMALIZE} | blank {BLANK_ID} | vocab {model.config.vocab_size}")
    return MODEL_DIR


def unload():
    global model, processor
    try:
        del model, processor
    except NameError:
        pass
    gc.collect()
    if DEVICE == "cuda":
        torch.cuda.empty_cache()
```

## 6 · Per-probe logging, and seeding the attack

Two additions over NB-21.

`pgd_batch_logged` returns, per sample, how many steps it used and why it
stopped — `success`, `stall` (the loss stopped improving by 1e-3 for
`STALL_PATIENCE` steps) or `step_cap`. The review asked for this because the
stall rule uses an **absolute** threshold, so how often it fires depends on each
model's loss scale. That is a run-level property that could masquerade as a
difference in attackability, and until it is logged nobody can tell.

`seed_attack` re-seeds Python, NumPy and Torch before each condition. The attack
starts from `delta = 0`, so the randomness that `ATTACK_SEED` actually moves is
cuDNN kernel selection, AMP loss scaling and any nondeterminism in the backward
pass — the same sources behind the 41%/61% pair.

```python
import random

def seed_attack(s):
    """Re-seed every generator, before each condition.

    Called for completeness and reproducibility of anything that might draw
    later. It does NOT make this run differ from the published one: the attack
    path initialises delta to zeros, uses no restarts and no dropout, and does
    not shuffle. The difference comes from cudnn kernel selection and fp16
    accumulation order. See the header cell.
    """
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if DEVICE == "cuda":
        torch.cuda.manual_seed_all(s)
    # Deliberately NOT setting deterministic algorithms: the published runs did
    # not, and matching them is the point. See NB-18 for the determinism control.


def pgd_batch_logged(x, mask, tgt_pad, tgt_len, targets, eps_vec,
                     steps, delta_init=None, lr=LR):
    """pgd_batch, plus per-sample steps_used and stop_reason.

    Identical optimisation to NB-21's pgd_batch -- same Adam, same clamp, same
    stall rule, same check interval. The only additions are the two bookkeeping
    arrays, which do not touch the gradient path.
    """
    B = x.shape[0]
    eps_col = eps_vec.view(-1, 1)
    delta = (torch.zeros_like(x) if delta_init is None
             else torch.max(torch.min(delta_init.clone(), eps_col), -eps_col))
    delta.requires_grad_(True)
    opt = torch.optim.Adam([delta], lr=lr)
    scaler = torch.amp.GradScaler("cuda", enabled=USE_AMP)

    active = torch.ones(B, dtype=torch.bool, device=DEVICE)
    frozen = torch.zeros_like(x)
    success = np.zeros(B, dtype=bool)
    best_loss = torch.full((B,), float("inf"), device=DEVICE)
    stall = torch.zeros(B, dtype=torch.long, device=DEVICE)

    steps_used = np.full(B, steps, dtype=int)       # NEW
    reason = np.array(["step_cap"] * B, dtype=object)  # NEW

    step = -1
    for step in range(steps):
        if not active.any():
            break
        x_adv = torch.clamp(x + delta, -1, 1) * mask
        with torch.amp.autocast("cuda", enabled=USE_AMP):
            per = ctc_per_sample(x_adv, mask, tgt_pad, tgt_len)
        loss = (per * active.float()).sum() / active.float().sum().clamp(min=1)

        opt.zero_grad(); scaler.scale(loss).backward()
        scaler.unscale_(opt)
        if delta.grad is not None:
            delta.grad[~active] = 0
        scaler.step(opt); scaler.update()
        with torch.no_grad():
            delta.data = torch.max(torch.min(delta.data, eps_col), -eps_col)

        with torch.no_grad():
            improved = per < best_loss - 1e-3
            best_loss = torch.minimum(best_loss, per.detach())
            stall = torch.where(improved, torch.zeros_like(stall), stall + 1)
            give_up = active & (stall > STALL_PATIENCE)
            if give_up.any():                        # NEW: record who stalled
                for i in torch.nonzero(give_up).flatten().tolist():
                    steps_used[i] = step + 1; reason[i] = "stall"
            active = active & ~give_up

        if (step + 1) % CHECK_EVERY == 0:
            hyp = decode_b(int16_rt(torch.clamp(x + delta, -1, 1)) * mask, mask)
            with torch.no_grad():
                for i in range(B):
                    if active[i] and hyp[i] == targets[i]:
                        frozen[i] = delta[i].detach()
                        success[i] = True
                        active[i] = False
                        steps_used[i] = step + 1; reason[i] = "success"   # NEW
    return success, frozen, step + 1, steps_used, reason


def eps50_batch_logged(x, mask, tgt_pad, tgt_len, targets, utt_ids):
    """eps50_batch with a probe-level log. Same bisection, same record() rule.

    Returns the usual (eps, width, confirmed, best_delta) plus a list of
    probe records: one dict per (utterance, probe).
    """
    B = x.shape[0]
    lo = torch.full((B,), EPS_LO, device=DEVICE)
    hi = torch.full((B,), EPS_HI, device=DEVICE)
    best_eps = torch.full((B,), float("nan"), device=DEVICE)
    best_delta = torch.zeros_like(x)
    log = []

    def record(okt, eps_vec, d):
        nonlocal best_eps, best_delta
        better = okt & (torch.isnan(best_eps) | (eps_vec < best_eps))
        best_eps = torch.where(better, eps_vec, best_eps)
        best_delta = torch.where(better.view(-1, 1), d, best_delta)

    def log_probe(name, eps_vec, ok, su, rs, budget):
        for i in range(B):
            log.append(dict(utt_id=utt_ids[i], probe=name,
                            eps=float(eps_vec[i]), success=bool(ok[i]),
                            steps_used=int(su[i]), step_budget=budget,
                            stop_reason=str(rs[i])))

    ok, d, _, su, rs = pgd_batch_logged(x, mask, tgt_pad, tgt_len, targets, hi,
                                        CONFIRM_STEPS)
    log_probe("ceiling", hi, ok, su, rs, CONFIRM_STEPS)
    okt = torch.tensor(ok, device=DEVICE)
    record(okt, hi, d)
    feasible = ok.copy()

    for rnd in range(BISECT_ROUNDS):
        mid = torch.sqrt(lo * hi)
        ok, d, _, su, rs = pgd_batch_logged(x, mask, tgt_pad, tgt_len, targets, mid,
                                            PROBE_STEPS, delta_init=best_delta)
        ok = ok & feasible
        log_probe(f"bisect{rnd+1}", mid, ok, su, rs, PROBE_STEPS)
        okt = torch.tensor(ok, device=DEVICE)
        hi = torch.where(okt, mid, hi)
        lo = torch.where(okt, lo, mid)
        record(okt, mid, d)

    mid = torch.sqrt(lo * hi)
    ok, d, _, su, rs = pgd_batch_logged(x, mask, tgt_pad, tgt_len, targets, mid,
                                        CONFIRM_STEPS, delta_init=best_delta)
    ok = ok & feasible
    log_probe("confirm", mid, ok, su, rs, CONFIRM_STEPS)
    okt = torch.tensor(ok, device=DEVICE)
    record(okt, mid, d)

    eps = best_eps.cpu().numpy()
    confirmed = ~np.isnan(eps)
    width = (hi / lo).cpu().numpy()
    return eps, width, confirmed, best_delta, log
```

## 7 · Per-condition sweep

```python
def run_condition(cfg, deadline):
    """One condition to completion or to the deadline. Writes after every batch.

    Two outputs per condition:
      attackrep_{tag}_a{seed}_T5_n100.csv        one row per utterance
      probes_{tag}_a{seed}_T5_n100.csv           one row per (utterance, probe)
    """
    tag = cfg["tag"]
    OUT = f"{WORK}/attackrep_{tag}_a{ATTACK_SEED}_{EDIT_TYPE}_n{N_EVAL}.csv"
    PRB = f"{WORK}/probes_{tag}_a{ATTACK_SEED}_{EDIT_TYPE}_n{N_EVAL}.csv"

    # Resume: this session's own output first, then any attached copy, at every
    # mount depth. A silent zero is the failure that matters, so say so loudly.
    results, probes = [], []
    _base = os.path.basename(OUT)
    srcs = [OUT] if os.path.exists(OUT) else []
    for _d in range(1, 5):
        srcs += sorted(glob.glob("/kaggle/input/" + "*/" * _d + _base))
    srcs = list(dict.fromkeys(srcs))
    seen = set()
    for p in srcs:
        _n = 0
        for r in pd.read_csv(p).to_dict("records"):
            if r["utt_id"] not in seen:
                seen.add(r["utt_id"]); results.append(r); _n += 1
        print(f"  resume source: +{_n:3d} new rows  {p}")
        pp = os.path.join(os.path.dirname(p), os.path.basename(PRB))
        if os.path.exists(pp):
            probes += pd.read_csv(pp).to_dict("records")
    if results:
        print(f"  resuming: {len(results)} utterances already measured")
    else:
        print(f"  NO RESUME SOURCE for {_base} -- starting from scratch.")
        print("    If you attached a previous partial run and expected to continue,")
        print("    stop and check the filename above appears in that dataset.")
    if len(results) >= len(ev):
        print(f"  ALREADY COMPLETE ({len(results)}/{len(ev)}) -- skipping")
        pd.DataFrame(results).to_csv(OUT, index=False)
        return "skipped", results

    todo = ev[~ev.utt_id.isin(seen)].reset_index(drop=True)
    batches = [todo.iloc[i:i+BATCH_SIZE] for i in range(0, len(todo), BATCH_SIZE)]
    print(f"  {len(todo)} remaining -> {len(batches)} batches of up to {BATCH_SIZE}")

    for bi, sub in enumerate(batches):
        if time.time() > deadline:
            print(f"  *** wall clock reached, {sum(len(b) for b in batches[bi:])} "
                  f"utterances unmeasured in this condition ***")
            return "partial", results
        xb, mb = load_batch(sub)
        tp = [torch.tensor(processor.tokenizer(t).input_ids, device=DEVICE)
              for t in sub.target_text]
        tl = torch.tensor([len(t) for t in tp], device=DEVICE)
        tpad = torch.nn.utils.rnn.pad_sequence(tp, batch_first=True, padding_value=0)
        t0 = time.time()
        eps, width, conf, best, log = eps50_batch_logged(
            xb, mb, tpad, tl, list(sub.target_text), list(sub.utt_id))
        dt = time.time() - t0
        x_adv = int16_rt(torch.clamp(xb + best, -1, 1)) * mb
        snrs = snr_db_b(xb, x_adv, mb)
        hyp = decode_b(x_adv, mb)
        for rec in log:
            rec.update(tag=tag, attack_seed=ATTACK_SEED)
        probes += log
        for i, (_, r) in enumerate(sub.iterrows()):
            ok = not np.isnan(eps[i])
            mine = [q for q in log if q["utt_id"] == r["utt_id"]]
            results.append(dict(
                utt_id=r["utt_id"], tag=tag, attack_seed=ATTACK_SEED, rung=RUNG,
                edit_type=EDIT_TYPE, checkpoint=CHECKPOINT, duration=r["duration"],
                source_text=r["text"], target_text=r["target_text"],
                source_tokens=int(r["source_tokens"]),
                target_tokens=int(r["target_tokens"]),
                eps50=float(eps[i]) if ok else None, censored=not ok,
                eps50_per_token=(float(eps[i]) / r["target_tokens"]) if ok else None,
                bracket_width=float(width[i]), snr_db=float(snrs[i]),
                decoded=(hyp[i] if ok else None),
                exact_match=(bool(hyp[i] == r["target_text"]) if ok else False),
                steps_total=int(sum(q["steps_used"] for q in mine)),
                n_stall=int(sum(q["stop_reason"] == "stall" for q in mine)),
                n_stepcap=int(sum(q["stop_reason"] == "step_cap" for q in mine)),
                use_amp=USE_AMP, model_dir=MODEL_DIR))
        pd.DataFrame(results).to_csv(OUT, index=False)
        pd.DataFrame(probes).to_csv(PRB, index=False)
        rem = len(batches) - (bi + 1)
        eta = time.time() + rem * dt
        print(f"    batch {bi+1}/{len(batches)}  {dt/60:4.1f} min  "
              f"{int((~np.isnan(eps)).sum())}/{len(sub)} uncensored  "
              f"{len(results)}/{len(ev)} rows"
              + (f"  | {rem} left, ~{rem*dt/3600:.1f} h"
                 f" ({'fits' if eta < deadline else 'WILL NOT FIT'})" if rem else ""))
    return ("complete" if len(results) >= len(ev) else "partial"), results
```

```python
SESSION_START = time.time()
DEADLINE = SESSION_START + WALL_HOURS * 3600
summary = []

for qi, cfg in enumerate(QUEUE, 1):
    left = (DEADLINE - time.time()) / 3600
    print("\n" + "=" * 74)
    print(f"[{qi}/{len(QUEUE)}] {cfg['tag']}  attack seed {ATTACK_SEED}"
          f"   ({left:.1f} h of wall clock left)")
    print("=" * 74)
    if left < MIN_HOURS_TO_START:
        print(f"  only {left:.1f} h left, below MIN_HOURS_TO_START={MIN_HOURS_TO_START}.")
        print("  NOT STARTING. Publish /kaggle/working, attach it back, re-commit.")
        summary.append(dict(tag=cfg["tag"], status="not-started", n=0, asr=None,
                            published_asr=cfg["published_asr"]))
        continue
    try:
        load_condition(cfg)
        seed_attack(ATTACK_SEED)          # re-seed AFTER the model is on the GPU
        status, rows = run_condition(cfg, DEADLINE)
    except Exception as e:
        print(f"  !! {type(e).__name__}: {e}")
        summary.append(dict(tag=cfg["tag"], status=f"error: {type(e).__name__}",
                            n=0, asr=None, published_asr=cfg["published_asr"]))
        unload(); continue
    df = pd.DataFrame(rows)
    asr = float((~df.censored.astype(bool)).mean() * 100) if len(df) else None
    summary.append(dict(tag=cfg["tag"], status=status, n=len(df), asr=asr,
                        published_asr=cfg["published_asr"]))
    print(f"  {status.upper()}  {len(df)}/{len(ev)} measured"
          + (f"  ASR@2e-3 = {asr:.1f}%  (first attack: {cfg['published_asr']:.1f}%,"
             f"  gap {abs(asr-cfg['published_asr']):.1f} pp)" if asr is not None else ""))
    unload()

print("\n" + "=" * 74)
print(f"SESSION DONE in {(time.time()-SESSION_START)/3600:.2f} h")
print("=" * 74)
```

## 9 · Attack variance, and the stop-reason audit

```python
import math

sm = pd.DataFrame(summary)
print(sm.to_string(index=False))

done = sm[(sm.status.isin(["complete", "skipped"])) & sm.asr.notna()]
print("\n--- attack variance at FIXED weights " + "-" * 36)
if len(done) == 0:
    print("  no condition finished yet; nothing to pool.")
else:
    diffs = []
    for r in done.itertuples():
        d = abs(r.asr - r.published_asr)
        diffs.append(d)
        print(f"  {r.tag:16s} {r.published_asr:5.1f}%  vs {r.asr:5.1f}%   "
              f"gap {d:5.1f} pp")
    print(f"\n  the pair we already had: 1 h traj weights, 41.0% vs 61.0%, 20.0 pp")
    allpairs = diffs + [20.0]
    # Each pair of measurements gives variance diff^2/2 on 1 df -- the same rule
    # the paper uses for the training-variance pairs.
    v = sum(d * d / 2 for d in allpairs) / len(allpairs)
    raw = math.sqrt(v)
    df_ = len(allpairs)
    c4 = math.exp(0.5 * math.log(2.0 / df_) + math.lgamma((df_ + 1) / 2.0)
                  - math.lgamma(df_ / 2.0))
    print(f"\n  ATTACK sd over {df_} pairs: {raw:.3f} pp raw, {raw/c4:.3f} pp "
          f"after the c4 correction (df {df_})")
    sigma_total = 21.574
    if raw / c4 < sigma_total:
        tr = math.sqrt(sigma_total ** 2 - (raw / c4) ** 2)
        print(f"  subtracted in quadrature from sigma = {sigma_total:.3f}: "
              f"{tr:.2f} pp left for TRAINING")
    else:
        print(f"  the attack component is NOT smaller than sigma = {sigma_total:.3f}. "
              "If that holds up, sigma is mostly measurement, and the paper's")
        print("  'train twice' prescription becomes 'attack twice'. Report it either way.")

print("\n--- why probes stopped " + "-" * 50)
pf = sorted(glob.glob(f"{WORK}/probes_*_a{ATTACK_SEED}_*.csv"))
if pf:
    pr = pd.concat([pd.read_csv(p) for p in pf], ignore_index=True)
    tab = pr.groupby(["tag", "stop_reason"]).size().unstack(fill_value=0)
    print(tab.to_string())
    print("\n  'stall' fires on an ABSOLUTE 1e-3 loss threshold, so its rate depends")
    print("  on each model's loss scale. A rate that differs a lot between rungs is")
    print("  a run-level property that would look like a difference in attackability.")
    print("\n  median steps used, by probe and stop reason:")
    print(pr.groupby(["probe", "stop_reason"]).steps_used.median().to_string())
else:
    print("  no probe files written yet.")

PROV.update(notebook=NOTEBOOK, queue=[c["tag"] for c in QUEUE],
            attack_seed=ATTACK_SEED, summary=summary, wall_hours=WALL_HOURS,
            session_hours=round((time.time()-SESSION_START)/3600, 2),
            n_eval=int(len(ev)), edit_type=EDIT_TYPE, seed=SEED,
            bisect_rounds=BISECT_ROUNDS, probe_steps=PROBE_STEPS,
            confirm_steps=CONFIRM_STEPS, batch_size=BATCH_SIZE, lr=LR,
            eps_lo=EPS_LO, eps_hi=EPS_HI, use_amp=USE_AMP)
json.dump(PROV, open(f"{WORK}/_provenance.json", "w"), ensure_ascii=False,
          indent=2, default=str)
json.dump({"title": f"attack-seed-replicates-a{ATTACK_SEED}",
           "id": f"YOURUSERNAME/attack-seed-replicates-a{ATTACK_SEED}",
           "licenses": [{"name": "CC0-1.0"}]},
          open(f"{WORK}/dataset-metadata.json", "w"), indent=2)
print(f"\nwrote {len([f for f in os.listdir(WORK) if f.endswith('.csv')])} CSVs to {WORK}")
for f in sorted(os.listdir(WORK)):
    print("   ", f)
print("\nCHECK THIS FILE LIST before publishing: publishing REPLACES the dataset and")
print("deletes anything this run did not regenerate.")
if any(s["status"] in ("partial", "not-started") for s in summary):
    print("\n*** NOT EVERYTHING FINISHED. Publish, attach the output back as an input,")
    print("    and re-commit: finished utterances are skipped, not redone. ***")
```

## What to do with the result

The number this produces is the **attack** standard deviation. The paper
currently reports sigma = 21.6 pp and reads it as an upper bound on training
variance, with the measurement component bounded loosely at 8.5–14 pp from a
single pair.

Two outcomes, and the paper is publishable either way:

**The attack component is small.** Then most of sigma really is training, the
title's prescription stands as written, and §6.4 can state the split instead of
bounding it.

**The attack component is large.** Then sigma is mostly measurement noise, and
the finding changes shape: targeted attack success on low-data ASR models cannot
be measured reliably with a single attack run. That is still a useful result,
and arguably a more actionable one — it is cheaper to re-attack than to retrain.

Either way, report the stop-reason table. If `stall` fires at very different
rates across rungs, say so: an absolute loss threshold on models with different
loss scales is a confound in its own right, independent of everything else.
